```python
#pip install plotly
import matplotlib as plt
import seaborn as sns
import plotly
import plotly.express as px
import pandas as pd
```

```python
data = sns.load_dataset('titanic')
df = pd.DataFrame(data)
df.head()
```

```python
df.groupby('sex').size().plot(kind='barh', color=sns.palettes.mpl_palette('Dark2'))
plt.gca().spines[['top', 'right',]].set_visible(True)
```

```python
df_grouped = df.groupby('sex').size().reset_index(name='count')

fig = px.bar(df_grouped, x='count', y='sex', orientation='h',
             title='Count by Sex', color='sex', color_discrete_sequence=px.colors.qualitative.Dark2)

fig.show()
```
